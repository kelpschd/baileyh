```python
# Helper function to pull metadata info from image

def get_channel_indices(row, n_channels=4):
    channel_map = {}

    for i in range(n_channels):
        name = row.get(f"ch{i}")
        if name:
            channel_map[name] = i
    return channel_map

def get_image_path(row, img_lookup):
    return img_lookup.get(row["filename"])
```

### ID lysosomes from a single image

```python
from scipy.ndimage import gaussian_laplace
from skimage import filters, morphology
from scipy import ndimage as ndi
import numpy as np
from scipy.stats import norm

# Paramaters - same as ACIS standard
intensity_scaling_param = [3, 19]
min_lysosome_area = 5
blur_sigma = 1
log_sigma_1, log_cutoff_1 = 3, 0.13
log_sigma_2, log_cutoff_2 = 2, 0.08
log_sigma_3, log_cutoff_3 = 1, 0.06
vesselness_sigma = [1]
vesselness_cutoff = 0.3

lys_ch = img[lyso_ch].copy().astype(float)

# Normalize to normal distribution
m, s = norm.fit(lys_ch.flatten())
stretch_min = max(m - intensity_scaling_param[0] * s, lys_ch.min())
stretch_max = min(m + intensity_scaling_param[1] * s, lys_ch.max())
lys_ch_n = np.clip(lys_ch, stretch_min, stretch_max)
image_norm = (lys_ch_n - stretch_min) / (stretch_max - stretch_min)

# Blur image
blurred = filters.gaussian(image_norm, sigma=blur_sigma)

# Identify blobs
log_1 = -1 * (log_sigma_1**2) * gaussian_laplace(blurred, sigma=log_sigma_1)
log_2 = -1 * (log_sigma_2**2) * gaussian_laplace(blurred, sigma=log_sigma_2)
log_3 = -1 * (log_sigma_3**2) * gaussian_laplace(blurred, sigma=log_sigma_3)

log_1_mask = log_1 > log_cutoff_1
log_2_mask = log_2 > log_cutoff_2
log_3_mask = log_3 > log_cutoff_3

log_mask = np.logical_or(log_1_mask, log_2_mask)
log_mask = np.logical_or(log_mask, log_3_mask)

# Identify vessels
vesselness = filters.frangi(blurred, sigmas=vesselness_sigma, black_ridges=False) > vesselness_cutoff
combined = np.logical_or(log_mask, vesselness)

filled = ndi.binary_fill_holes(combined)
lysosome_mask = morphology.remove_small_objects(filled, min_size=min_lysosome_area)
```

## Testing new masking approach

```python
from scipy.ndimage import gaussian_laplace
from skimage import filters, morphology
from scipy import ndimage as ndi
import numpy as np
from scipy.stats import norm

# Paramaters - same as ACIS standard
intensity_scaling_param = [3, 19]
blur_sigma = 1
log_sigma = 2
min_area = 5

out_img_list = []
out_name_list = []

lys_ch = img[lyso_ch].copy().astype(float)

# Normalize image
#image_norm = (nuc - np.min(nuc)) / (np.max(nuc) - np.min(nuc))

# Normalize to normal distribution
m, s = norm.fit(lys_ch.flatten())
stretch_min = max(m - intensity_scaling_param[0] * s, lys_ch.min())
stretch_max = min(m + intensity_scaling_param[1] * s, lys_ch.max())
lys_ch_n = np.clip(lys_ch, stretch_min, stretch_max)
image_norm = (lys_ch_n - stretch_min) / (stretch_max - stretch_min)

# Blur image
blurred = filters.gaussian(image_norm, sigma=blur_sigma)

# step 1: low level thresholding
triangle_cutoff = filters.threshold_triangle(blurred)
global_median_cutoff = np.percentile(blurred, 50)

th_low_cutoff = (triangle_cutoff + global_median_cutoff) / 2
img_low_level = blurred > th_low_cutoff

img_low_level_small = morphology.remove_small_objects(img_low_level, min_size=min_area, connectivity=1, out=img_low_level)
img_low_level_small_grow = morphology.dilation(img_low_level_small, footprint=morphology.disk(2))

# step 2: high level thresholding
otsu_cutoff = 0.333 * filters.threshold_otsu(blurred)
img_high_level = np.zeros_like(img_low_level_small_grow)
lab_low, num_obj = morphology.label(img_low_level_small_grow, return_num=True, connectivity=1)
for idx in range(num_obj):
    single_obj = lab_low == (idx + 1)
    local_otsu = filters.threshold_otsu(blurred[single_obj])
    if local_otsu > otsu_cutoff:
        img_high_level[np.logical_and(blurred > 0.98 * local_otsu, single_obj)] = 1

out_img_list.append(img_high_level.copy())
out_name_list.append("bw_coarse")

log_response = -1 * (log_sigma**2) * gaussian_laplace(blurred, sigma=log_sigma)

bw_extra = log_response > 0.07  # dot_2d_cutoff
bw_extra[~img_low_level_small_grow] = 0

bw_final = np.logical_or(bw_extra, img_high_level)

filled = ndi.binary_fill_holes(bw_final)

labeled_filled = morphology.label(filled, connectivity=1)
lysosome_mask = morphology.remove_small_objects(labeled_filled, min_size=min_area)
lysosome_mask = lysosome_mask > 0  # binary mask again

print(otsu_cutoff)

fig, axs = plt.subplots(1, 9, figsize=(20, 5))
fig.suptitle("MODIFIED AICS SEGMENTATION OF NPM1")

axs[0].imshow(lys_ch, cmap='gray') 
axs[0].set_title("DAPI (405) - NUCLEUS")
axs[0].axis("off")

axs[1].imshow(image_norm, cmap='gray')
axs[1].set_title("NORMALIZED IMAGE")
axs[1].axis("off")

axs[2].imshow(blurred, cmap='gray')
axs[2].set_title("BLURRED")
axs[2].axis("off")

axs[3].imshow(img_low_level)
axs[3].set_title("LOW LEVEL THRESHOLD")
axs[3].axis("off")

axs[4].imshow(img_low_level_small)
axs[4].set_title("LOW LEVEL THRESHOLD")
axs[4].axis("off")

axs[5].imshow(img_low_level_small_grow)
axs[5].set_title("LOW LEVEL THRESHOLD")
axs[5].axis("off")

axs[6].imshow(img_high_level)
axs[6].set_title("HIGH LEVEL THRESHOLD")
axs[6].axis("off")

axs[7].imshow(bw_final)
axs[7].set_title("FINAL")
axs[7].axis("off")

axs[8].imshow(lysosome_mask)
axs[8].set_title("FINAL CLEANED")
axs[8].axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy import ndimage as ndi
from skimage.measure import label, regionprops_table
import pandas as pd
 
cell_ids = np.unique(cellpose_cell_masks)
cell_ids = cell_ids[cell_ids != 0]

lys_labels_global = np.zeros_like(lysosome_mask, dtype=np.int32)
prab_labels_global = np.zeros_like(prab10_mask, dtype=np.int32)

lys_parent_cell = {}
prab_parent_cell = {}

lys_current_label = 1
prab_current_label = 1

for cid in cell_ids:
    single_cell_mask = (cellpose_cell_masks == cid)

    lys_in_cell = (lysosome_mask > 0) & single_cell_mask
    prab_in_cell = (prab10_mask > 0) & single_cell_mask

    lys_labeled_in_cell, lys_n = ndi.label(lys_in_cell)
    prab_labeled_in_cell, prab_n = ndi.label(prab_in_cell)

    if lys_n > 0:
        for ll in np.unique(lys_labeled_in_cell)[1:]:
            lys_labels_global[lys_labeled_in_cell == ll] = lys_current_label
            lys_parent_cell[lys_current_label] = int(cid)
            lys_current_label += 1

    # write prab10 objects into global label image + mapping
    if prab_n > 0:
        for ll in np.unique(prab_labeled_in_cell)[1:]:
            prab_labels_global[prab_labeled_in_cell == ll] = prab_current_label
            prab_parent_cell[prab_current_label] = int(cid)
            prab_current_label += 1

lys_int = img[lyso_ch] 
prab_int = img[prab10_ch] 

# We can consider changes to the regtion props table - think of this as similar to the measurments window in FIJI
# see documentation for more info!
# https://scikit-image.org/docs/0.25.x/api/skimage.measure.html#skimage.measure.regionprops_table

# Lysosomes
lys_props = regionprops_table(
    lys_labels_global,
    intensity_image=lys_int,
    properties=("label", "area", "min_intensity", "mean_intensity", "max_intensity")
)
lys_df = pd.DataFrame(lys_props)
lys_df["cell_id"] = lys_df["label"].map(lys_parent_cell)

# PRAB10
prab_props = regionprops_table(
    prab_labels_global,
    intensity_image=prab_int,
    properties=("label", "area", "min_intensity", "mean_intensity", "max_intensity")
)
prab_df = pd.DataFrame(prab_props)
prab_df["cell_id"] = prab_df["label"].map(prab_parent_cell)

lys_df.insert(0, "image_id", os.path.basename(img_path))
prab_df.insert(0, "image_id", os.path.basename(img_path))

print(lys_df)
print(prab_df)
```

## Lysosome and pRab10 colocalization data extraction

```python
import numpy as np
import pandas as pd
from skimage.filters import threshold_otsu
import os

lys = img[lyso_ch].astype(np.float64)
prab = img[prab10_ch].astype(np.float64)

# --- Global totals (per image) ---
cell_union = cellpose_cell_masks > 0
total_cell_mask_area_px = int(cell_union.sum())

# global (whole-image) Otsu thresholds (alternative: use your per-cell thresholds, but then “total segmented area”
# becomes ambiguous; global thresholds make the definition consistent)
try:
    tA_global = threshold_otsu(lys) if np.unique(lys).size > 1 else 0.0
except Exception:
    tA_global = 0.0
try:
    tB_global = threshold_otsu(prab) if np.unique(prab).size > 1 else 0.0
except Exception:
    tB_global = 0.0

lys_seg = lys > tA_global
prab_seg = prab > tB_global

segmented_area_lys_px = int(lys_seg.sum())
segmented_area_prab_px = int(prab_seg.sum())

# Often useful: segmented area only inside cells
segmented_area_lys_within_cells_px = int((lys_seg & cell_union).sum())
segmented_area_prab_within_cells_px = int((prab_seg & cell_union).sum())

# --- Per-cell coloc + per-cell segmented areas ---
cell_ids = np.unique(cellpose_cell_masks)
cell_ids = cell_ids[cell_ids != 0]

rows = []
min_pixels = 50  # skip tiny masks

for cid in cell_ids:
    m = (cellpose_cell_masks == cid)
    npx = int(m.sum())
    if npx < min_pixels:
        continue

    a = lys[m]
    b = prab[m]

    # Pearson r (guard constant arrays)
    if a.std() == 0 or b.std() == 0:
        pearson_r = np.nan
    else:
        pearson_r = float(np.corrcoef(a, b)[0, 1])

    # per-cell Otsu thresholds (your existing approach)
    try:
        tA = threshold_otsu(a) if np.unique(a).size > 1 else 0.0
    except Exception:
        tA = 0.0
    try:
        tB = threshold_otsu(b) if np.unique(b).size > 1 else 0.0
    except Exception:
        tB = 0.0

    a_pos = a > tA
    b_pos = b > tB

    denomA = a[a_pos].sum()
    denomB = b[b_pos].sum()

    M1 = float(a[a_pos & b_pos].sum() / denomA) if denomA > 0 else np.nan  # lys in prab
    M2 = float(b[b_pos & a_pos].sum() / denomB) if denomB > 0 else np.nan  # prab in lys

    # per-cell segmented areas (counts of above-threshold pixels within the cell)
    lys_seg_area_px_in_cell = int(a_pos.sum())
    prab_seg_area_px_in_cell = int(b_pos.sum())

    rows.append({
        "cell_id": int(cid),
        "cell_mask_area_px": npx,  # same as your cell_pixels
        "pearson_r": pearson_r,
        "manders_M1_lys_in_prab": M1,
        "manders_M2_prab_in_lys": M2,
        "threshold_cutoff_lys_cell": float(tA),
        "threshold_cutoff_prab_cell": float(tB),
        "lys_seg_area_px_in_cell": lys_seg_area_px_in_cell,
        "prab_seg_area_px_in_cell": prab_seg_area_px_in_cell,
    })

cell_coloc_df = pd.DataFrame(rows)
cell_coloc_df.insert(0, "image_id", os.path.basename(img_path))

# Add global totals as columns repeated on every row (easy to group later),
# or put them in a separate per-image dataframe if you prefer.
cell_coloc_df["total_cell_mask_area_px"] = total_cell_mask_area_px
cell_coloc_df["segmented_area_lys_px_global"] = segmented_area_lys_px
cell_coloc_df["segmented_area_prab_px_global"] = segmented_area_prab_px
cell_coloc_df["segmented_area_lys_within_cells_px_global"] = segmented_area_lys_within_cells_px
cell_coloc_df["segmented_area_prab_within_cells_px_global"] = segmented_area_prab_within_cells_px

print(cell_coloc_df)
```
